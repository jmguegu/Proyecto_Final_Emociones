# Análisis Exploratorio de Datos

En esta sección se realizará un Análisis Exploratorio de los Datos, donde se buscarán patrones, tendencias y correlaciones entre las variables.

Se utilizarán técnicas de visualización para facilitar la comprensión de los datos.

```python
# Importación de librerías y configuración básica
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import warnings

warnings.simplefilter(action='ignore', category=UserWarning)
warnings.simplefilter(action='ignore', category=FutureWarning)
warnings.simplefilter(action='ignore', category=pd.errors.SettingWithCopyWarning)
```

```python
data_basepath = '..\\data\\spain_dataframes\\'
```

```python
# Leer el fichero con los datos
df = pd.read_csv(f'{data_basepath}cleaned_data.csv')
df.describe()
```

```python
# Información general
print("Información general del DataFrame:")
print(df.info())
print("\nPrimeras filas del DataFrame:")
display(df.head())
```

```python
# Estadísticas descriptivas para variables numéricas
print("\nEstadísticas descriptivas (variables numéricas):")
display(df.describe())
```

```python
# Estadísticas descriptivas para variables categóricas
print("\nValores únicos por columna categórica:")
for col in df.select_dtypes(include='object').columns:
    print(f"{col}: {df[col].nunique()} únicos - {df[col].unique()[:5]}...")
```

```python
# Valores nulos
print("\nConteo de valores nulos por columna:")
print(df.isnull().sum())
```

```python
# Distribución de variables numéricas
num_cols = df.select_dtypes(include=[np.number]).columns
df[num_cols].hist(figsize=(15, 8), bins=30)
plt.suptitle('Distribución de variables numéricas')
plt.tight_layout()
# plt.savefig(f'{img_basepath}numeric.png', dpi=300, bbox_inches='tight')
plt.show()
```

### Distribución de variables numéricas

#### Edad

La distribución de la edad se corresponde con una población envejecida, con un pico en la franja de 60 a 70 años. La media de la edad es de 47 años, con una desviación estándar de 20 años. Esto indica que la mayoría de los pacientes tienen entre 37 y 67 años.

### Días alta contaminación

La variable "Días alta contaminación" muestra una distribución sesgada a la izquierda, con un pico en los 5 días. La media es de 4 días, con una desviación estándar de 2 días. Esto sugiere que la mayoría de los pacientes experimentan entre 2 y 6 días de alta contaminación.

Las comunidades con mayor número de días de alta contaminación se corresponde con zonas del sur de España, que sugiere que, con el parámetro de medición de la contaminación del aire usado, los días que se supera una concentración mayor de 50 microgramos por metro cúbico de partículas PM10, se dan típicamente en zonas con aire procedente de África, como es el caso de Andalucía y Murcia.

### Horas trabajadas al mes y salario anual

Ambas siguen una distribución normal, con una media en 140 h/mes y 24000 €/año . El pico de 0 horas trabajadas y salario de 0€ se debe al número de menores de 16 años que no trabajan, que provoca una distorsión en la distribución.

### Percepción de seguridad

Distribución uniforme, con una nota media de prácticamente 7, y una desviación estándar de 0.8, por lo que indica que la percepción de seguridad general de los datos es consistente, y se relaciona con una población que se siente segura en su entorno.

### Nivel de felicidad

Vemos un valle en torno a los 6.8, lo que indica que la sensación de felicidad pasa de una nota "regular" a una nota "buena" a partir de los 7.5, aún siendo la media de 7.1, con una desviación estándar de 1.1. La tendencia es que la población se siente feliz, pero con una ligera insatisfacción.

```python
# Distribución de variables categóricas
for col in df.select_dtypes(include='object').columns:
    plt.figure(figsize=(6,3))
    df[col].value_counts().plot(kind='bar', color=plt.cm.viridis(np.linspace(0, 1, df[col].nunique())))
    plt.title(f'Distribución de {col}')
    plt.ylabel('Frecuencia')
    plt.xlabel(col)
    plt.tight_layout()
    plt.show()
```

## Distribución de variables categóricas

### Distribución de género

Son muy igualadas, con un valor aproximado de hombres y mujeres, con una ligera mayoría de mujeres (51.5%).

### Distribución de actividad física

La mayoría de la población tiene un nivel bajo de actividad física. La población se considera sedentaria.

### Distribución de asistencia a eventos

En todos los casos (cine, deporte, eventos culturales y directos), la mayoría de la población no asiste, siendo la razón otra que la falta de recursos (no_puede), por lo que se entiende que no hay mucho interés general en este tipo de actividades.

### Distribución de estudios

La población se reparte mayoritariamente entre estudios secundarios y universitarios, con una muy baja tasa de analfabetismo. Si relacionamos la distribución de estudios con la de edades, vemos que la mayoría de la mayoría de la población en edad de estudiar, continúa sus estudios hasta la universidad.

### Distribución de estado civil

La mayoría de la población se encuentra casada, o soltera, con una muy baja proporción de viudos y divorciados. Esto sugiere una población estable en cuanto a relaciones personales.

### Distribución de satisfacción sanitaria

A todos los niveles (hospitales, dentistas, especialidades y medicina general), la población está satisfecha o muy satisfecha (en menor medida) en términos generales. Tanto las categorías de "insatisfecho" como "muy insatisfecho" son muy bajas, lo que indica una percepción equilibrada entre extremamente positiva y extremamente negativa, siendo la positiva la más predominante.

### Distribución de la frecuencia de felicidad

La población ha sentido momentos de felicidad en el último mes muy a menudo, lo que indique que es generalmente feliz, o al menos siente momentos de felicidad con frecuencia.

La poca población que indica que nunca ha tenido un momento de felicidad en el último mes es muy baja, por lo que lo consideraremos poco relevante para el análisis.

```python
# Correlación entre variables numéricas
corr = df[num_cols].corr()
plt.figure(figsize=(10,8))
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Matriz de correlación')
plt.show()
```

## Correlación entre variables numéricas

Exceptuando algunas variables como días de alta contaminación, horas trabajadas al mes y salario anual, la mayoría de las variables numéricas no presentan una correlación significativa entre sí.

Vemos que los días de alta contaminación sí tiene una ligera correlación negativa con el nivel de felicidad, lo que sugiere que a mayor número de días de alta contaminación, menor es la percepción de felicidad de la población, aunque esta se baja.

Las horas trabajadas al mes y el salario anual son las variagles con mayor relación entre ellas, como es de esperar, ya que a mayor número de horas trabajadas al mes se presupone un mayor salario anual. Ambas tienen una ligera correlación positiva con la edad, lo que sugiere que a medida que la población envejece, trabaja más horas al mes y tiene un mayor salario anual.
